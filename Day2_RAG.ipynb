## Important Note on Chunking:
A text chunk created from a TXT file will NOT contain PDF text content, and vice versa.

TXT file  ──► Document A ──► chunks A1, A2

PDF page ──► Document B ──► chunks B1, B2

DOCX     ──► Document C ──► chunks C1, C2

✔ Chunks never mix content from different documents

✔ Overlap happens only within the same document


```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=2500,
    chunk_overlap=1000
)

chunks = text_splitter.split_documents(all_documents)

print(f"Total chunks created: {len(chunks)}")
```

```python
from langchain_community.document_loaders import TextLoader
from langchain_core.documents import Document

md_docs = TextLoader("sample_data/README.md", encoding="utf-8").load()
txt_docs = TextLoader("sample.txt", encoding="utf-8").load()
```

```python
from langchain_text_splitters import MarkdownHeaderTextSplitter

md_splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=[
        ("#", "h1"),
        ("##", "h2"),
        ("###", "h3")
    ]
)

md_chunks = []

for doc in md_docs:
    splits = md_splitter.split_text(doc.page_content)
    for split in splits:
        md_chunks.append(
            Document(
                page_content=split.page_content,
                metadata={**doc.metadata, **split.metadata}
            )
        )
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

txt_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

txt_chunks = txt_splitter.split_documents(txt_docs)
```

```python
all_chunks = md_chunks + txt_chunks

print(f"Markdown chunks: {len(md_chunks)}")
print(f"Text chunks: {len(txt_chunks)}")
print(f"Total chunks: {len(all_chunks)}")
```

```python
example = all_chunks[0]

print(example.page_content[:300])
print(example.metadata)
```

## Embeddings

```python
!pip install langchain-openai
```

```python
phrases = [
    "apple",
    "mat",
    "math",
    "statistics"
]

query = "math"
```

```python
import os
import getpass

if "OPENAI_API_KEY" not in os.environ:
    os.environ["OPENAI_API_KEY"] = getpass.getpass("Enter your OpenAI API key: ")
```

```python
from langchain_openai import OpenAIEmbeddings
import numpy as np

embeddings = OpenAIEmbeddings()

phrase_embeddings = embeddings.embed_documents(phrases)
query_embedding = embeddings.embed_query(query)

print(phrase_embeddings)
print(query_embedding)


```

```python
def cosine_sim(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

scores = [
    cosine_sim(query_embedding, emb)
    for emb in phrase_embeddings
]

print("\nEmbedding similarity scores:")
for phrase, score in zip(phrases, scores):
    print(f"{phrase:12s} -> {score:.3f}")
```

```python
import os
import getpass

if "OPENAI_API_KEY" not in os.environ:
    os.environ["OPENAI_API_KEY"] = getpass.getpass("Enter OpenAI API Key: ")
```

```python
from langchain_core.documents import Document

documents = [
    Document(page_content="Math is the study of numbers and equations."),
    Document(page_content="Statistics focuses on data analysis and probability."),
    Document(page_content="A cat is a small domesticated animal."),
    Document(page_content="A mat is an object placed on the floor.")
]
```

```python
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings()
```

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(
    documents=documents,
    embedding=embeddings
)
```

```python
query = "What is related to mathematics?"

results = vectorstore.similarity_search(query, k=2)

for doc in results:
    print("-", doc.page_content)
```
